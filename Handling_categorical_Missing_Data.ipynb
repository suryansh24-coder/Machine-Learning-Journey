```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('train.csv' ,usecols=['GarageQual','FireplaceQu','SalePrice'])
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
df['GarageQual'].value_counts().sort_values(ascending=False).plot.bar()
plt.xlabel("GarageQual")
plt.ylabel("Count")
```

```python
df['GarageQual'].mode()
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

df[df['GarageQual']=='TA']['SalePrice'].plot(kind='kde',ax=ax)
df[df['GarageQual'].isnull()]['SalePrice'].plot(kind='kde',ax=ax,color='red')

lines , labels = ax.get_legend_handles_labels()
labels = ['Houses with TA','Houses with NA']
ax.legend(lines , labels ,loc='best')
plt.title("GarageQual")
```

```python
temp = df[df['GarageQual'] == 'TA']['SalePrice']
```

```python
df['GarageQual'].fillna('TA',inplace=True)
```

```python
df['GarageQual'].value_counts().plot(kind='bar')
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)
temp.plot(kind='kde' ,ax=ax)

df[df['GarageQual'] == 'TA']['SalePrice'].plot(kind='kde',ax=ax,color='red')

lines,labels = ax.get_legend_handles_labels()
labels = ["Original Variable" , "Imputed Variables"]
ax.legend(lines ,labels ,loc='best')

plt.title("GarageQual")
plt.grid(True)
```

```python
df['FireplaceQu'].value_counts().plot(kind='bar')
```

```python
df['FireplaceQu'].mode()
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

df[df['FireplaceQu']=='Gd']['SalePrice'].plot(kind='kde',ax=ax)
df[df['FireplaceQu'].isnull()]['SalePrice'].plot(kind='kde',ax=ax,color='red')

lines, labels = ax.get_legend_handles_labels()
labels = ['Houses with Gd','Houses with NA']
ax.legend(lines , labels ,loc='best')
plt.title("FireplaceQu")
plt.grid()
```

```python
temp = df[df['FireplaceQu'] == 'Gd']['SalePrice']
```

```python
df['FireplaceQu'].fillna('Gd' , inplace=True)
```

```python
df['FireplaceQu'].value_counts().plot(kind='bar')
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

temp.plot(kind='kde',ax=ax) # Plotting original 'Gd' distribution
df[df['FireplaceQu']=='Gd']['SalePrice'].plot(kind='kde',ax=ax,color='red') # Plotting combined 'Gd' (original + imputed)

lines, labels = ax.get_legend_handles_labels()
labels = ['Original Variable','Imputed Variables'] # Updated labels for clarity
ax.legend(lines , labels ,loc='best')
plt.title("FireplaceQu")
plt.grid()
```
