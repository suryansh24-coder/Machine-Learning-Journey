```python
from sklearn.datasets import load_diabetes

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from sklearn.linear_model import Lasso
from sklearn.metrics import r2_score

from sklearn.model_selection import train_test_split
```

```python
data = load_diabetes()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop('target', axis=1), df['target'], test_size=0.2, random_state=2)
```

```python
from sklearn.metrics import r2_score

if 'r2_score' in globals() and isinstance(globals()['r2_score'], list):
    del r2_score

coefs = []
r2_scores = []

for i in [0, 0.1, 1, 10]:
  reg = Lasso(alpha=i)
  reg.fit(X_train, y_train)
  coefs.append(reg.coef_.tolist())
  y_pred = reg.predict(X_test)
  r2_scores.append(r2_score(y_test, y_pred))
```

```python
plt.figure(figsize=(14, 9))
plt.subplot(221)
plt.bar(data.feature_names , coefs[0])
plt.title('Alpha = 0 ,r2_score={}'.format(round(r2_scores[0],2)))

plt.subplot(222)
plt.bar(data.feature_names , coefs[1])
plt.title('Alpha = 0.1 ,r2_score={}'.format(round(r2_scores[1],2)))

plt.subplot(224)
plt.bar(data.feature_names , coefs[3])
plt.title('Alpha = 10 ,r2_score={}'.format(round(r2_scores[3],2)))

plt.show()
```

```python
# Impact on Bias and Variance :-
import numpy as np
import matplotlib.pyplot as plt

m = 100
X = 5 * np.random.rand(m,1) - 2
y = 0.7 * X ** 2 - 2 - X + 3 + np.random.rand(m,1)

plt.scatter(X,y)
plt.grid(True)
plt.show()
```

```python
X_train , X_test , y_train , y_test = train_test_split(X , y , test_size=0.2 , random_state=2)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=10)

X_train = poly.fit_transform(X_train)
X_test = poly.transform(X_test)
```

```python
from mlxtend.evaluate import bias_variance_decomp

alphas = np.linspace(0, 30, 100)
loss = []
bias = []
variance = []

for i in alphas:
    reg = Lasso(alpha=i)
    avg_expected_loss, avg_bias, avg_var = bias_variance_decomp(
        reg, X_train, y_train, X_test, y_test, loss='mse', random_seed=123
    )
    loss.append(avg_expected_loss)
    bias.append(avg_bias)
    variance.append(avg_var)
```

```python
plt.plot(alphas,loss,label='loss')
plt.plot(alphas,bias,label='Bias')
plt.plot(alphas,variance,label='Variance')
plt.xlabel('Alpha')
plt.grid(True)
plt.legend()
```
