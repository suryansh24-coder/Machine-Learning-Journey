```python
import pandas as pd
data = pd.read_csv("https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv")
data.head()
```

```python
x = data.drop('Outcome',axis=1)
y = data['Outcome']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_scores = model.predict_proba(x_test)[:,1]
```

```python
y_scores
```

```python
from sklearn.metrics import roc_auc_score, roc_curve
roc_auc_score(y_test, y_scores)
fpr, tpr, thresholds = roc_curve(y_test, y_scores)
```

```python
thresholds
```

```python
import plotly.graph_objects as go
fig = go.Figure()
fig.add_trace(go.Scatter(x=fpr, y=tpr, mode='lines'))
fig
```

```python
import numpy as np
trace0 = go.Scatter(
    x=fpr,
    y=tpr,
    mode='lines',
    name='ROC Curve'
)
n=10
indices = np.arange(len(thresholds)) % n == 0

trace1 = go.Scatter(
    x=fpr[indices],
    y=tpr[indices],
    mode='markers',
    name='Thresholds'
)

trace2 = go.Scatter(
    x=[0,1],
    y=[0,1],
    mode='lines',
    name='Random Classifier'
)
fig = go.Figure()
fig.add_trace(trace0)
fig.add_trace(trace1)
fig.add_trace(trace2)
fig
```

```python
potimal_idx = np.argmax(tpr - fpr)
potimal_threshold = thresholds[potimal_idx]
potimal_threshold
```

```python
import plotly.graph_objects as go
fig = go.Figure()
fig.add_trace(go.Scatter(x=fpr, y=tpr, mode='lines'))
fig.add_trace(go.Scatter(x=[0,1], y=[0,1], mode='lines'))
fig.add_trace(go.Scatter(x=[potimal_threshold], y=[1], mode='markers', marker=dict(color='red')))
fig
```
