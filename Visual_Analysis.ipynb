```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("50_Startups.csv")
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().mean()*100
```

```python
df.head()
```

```python
X = df['R&D Spend']
y = df['Marketing Spend']
size = df['Marketing Spend']
sns.scatterplot(x=X, y=y , markers='O' , edgecolor='blue' , facecolor= 'cyan' , alpha=0.5 , size=size)
plt.xlabel('R&D Spend')
plt.ylabel('Marketing Spend')
plt.title('R&D Spend vs Marketing Spend')
plt.grid(True)
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True) , annot=True)
plt.show()
```

```python
X = df['R&D Spend']
y = df['Marketing Spend']
sns.boxplot(X)
sns.boxplot(y)
plt.show()
plt.show()
```

```python
sns.pairplot(df)
```

```python
sns.kdeplot(df['Profit'])
sns.histplot(df['Profit'] , color='green' , kde=True)
plt.title("Profit Visualization")
plt.grid(True)
plt.show()
```

```python
df.head()
```

```python
sns.histplot(data=df, x='Profit', hue='State', kde=True, palette='viridis')
plt.title('Distribution of Profit by State')
plt.xlabel('Profit')
plt.ylabel('Count / Density')
plt.grid(True)
plt.show()
```
